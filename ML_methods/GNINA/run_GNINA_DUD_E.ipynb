```python
import pandas as pd
index = pd.read_csv("/content/drive/My Drive/DUD-E/index.csv");
index
```

```python
import os


# Paths
base_folder = "/content/drive/My Drive/DUD-E/"
output_folder = os.path.join(base_folder, "result_docking")

# Ensure output directory exists
os.makedirs(output_folder, exist_ok=True)

# Extract dataset columns
ligand_paths = index["ligandpath"]
protein_paths = index["proteinpath"]
system_names = index["systemname"]

for i, (lig_path, prot_path, name) in enumerate(zip(ligand_paths, protein_paths, system_names)):

    print(f"Processing system {i}: {name}")

    receptor = os.path.join(base_folder, prot_path)
    ligand = os.path.join(base_folder, lig_path)

    output_file = os.path.join(output_folder, f"{i}_{name}.sdf")

    command = f"./gnina -r '{receptor}' -l '{ligand}' --autobox_ligand '{ligand}' -o '{output_file}' --seed 0 "
    !{command}
```
